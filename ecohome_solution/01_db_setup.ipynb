# EcoHome database setup
Create the SQLite schema and seed seven days of usage history plus forecast data for the next week.

```python
from models.energy import initialize_database

# Create or reset the local sample database before the later setup steps.
database_path = initialize_database()
print(f"Initialized sample energy data: {database_path}")
```

```python
import sqlite3

# Review table sizes and device totals to confirm the seed data is present.
with sqlite3.connect(database_path) as connection:
    for table in ("energy_readings", "solar_generation", "electricity_prices", "weather_forecasts"):
        count = connection.execute(f"SELECT COUNT(*) FROM {table}").fetchone()[0]
        print(f"{table}: {count:,} rows")
    print("\nRecent device totals:")
    for row in connection.execute("SELECT device, ROUND(SUM(consumption_kwh), 1), ROUND(SUM(cost_usd), 2) FROM energy_readings GROUP BY device ORDER BY SUM(consumption_kwh) DESC"):
        print(f"{row[0]:12} {row[1]:>7} kWh  ${row[2]:.2f}")
```
