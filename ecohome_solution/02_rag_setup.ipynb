# Energy knowledge base
Index the local energy documents in Chroma and inspect the passages retrieved for an EV scheduling question. The deterministic feature-hash embeddings run locally and do not download a model.

```python
from tools import create_rag_index, retrieve_energy_tips

# Rebuild the local index from the current text files before querying it.
document_count = create_rag_index()
print(f"Indexed {document_count} knowledge-base passages.")
```

```python
# Retrieve the most relevant passages and print their source filenames for citation.
query = "How can I schedule EV charging around rooftop solar and time-of-use prices?"
for result in retrieve_energy_tips(query, top_k=3):
    print(f"[{result['source']}] {result['text']}\n")
```
